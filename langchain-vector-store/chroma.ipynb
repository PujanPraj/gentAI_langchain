```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_core.documents import Document

doc1 = Document(
    page_content="Cristiano Ronaldo is known for his goal-scoring ability, powerful shots, and athleticism.",
    metadata={"team": "Al Nassr"}
)

doc2 = Document(
    page_content="Lionel Messi is famous for his dribbling, playmaking, and exceptional ball control.",
    metadata={"team": "Inter Miami"}
)

doc3 = Document(
    page_content="Kylian Mbappe is known for his incredible speed, finishing, and attacking skills.",
    metadata={"team": "Real Madrid"}
)

doc4 = Document(
    page_content="Erling Haaland is a powerful striker known for his positioning and goal-scoring ability.",
    metadata={"team": "Manchester City"}
)

doc5 = Document(
    page_content="Mohamed Salah is known for his pace, dribbling, and ability to score goals from the right wing.",
    metadata={"team": "Liverpool"}
)
```

```python
documents = [doc1, doc2, doc3, doc4, doc5]
```

```python
embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vector_store = Chroma(
    embedding_function=embedding,
    collection_name="sample",
    persist_directory='chroma_db'
)
```

```python
vector_store.add_documents(documents)
```

```python
vector_store.get(include=["embeddings","documents", "metadatas"])
```

```python
vector_store.similarity_search(
    query="Who have the powerful shot",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="Who have the powerful shot",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Liverpool"}
)
```

```python
updated_doc1 = Document(
    page_content="Cristiano Ronaldo is a Portuguese footballer known for his incredible goal-scoring ability and athleticism.",
    metadata={"team": "Al Nassr"}
)

vector_store.update_document(document_id="de3ec31e-8b18-4d63-8f26-ede621bfa047", document=updated_doc1)
```

```python
vector_store.get(include=["embeddings", "documents", "metadatas"])
```
